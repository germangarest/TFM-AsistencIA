# Exploración y Visualización de Dataset de Peleas (YOLOv8)

Este notebook permite explorar y visualizar un dataset de peleas en formato YOLOv8 descargado desde Google Drive.

Dataset de roboflow: https://universe.roboflow.com/alexander-genza/v_d-jikoi

## 1. Descargar y extraer el dataset

Primero, montamos Google Drive y descargamos el archivo ZIP:

```python
from google.colab import drive
import gdown
import os
import zipfile

# Montar Google Drive
drive.mount('/content/drive')

# URL del archivo ZIP en Google Drive
url = 'https://drive.google.com/file/d/1D-5wYndw5VJItKBDXdc9ws2CBeHfQo15/view?usp=sharing'

# Convertir URL de Google Drive a formato compatible con gdown
file_id = url.split('/')[-2]
gdown_url = f'https://drive.google.com/uc?id={file_id}'

# Descargar el archivo
output_path = '/content/dataset.zip'
gdown.download(gdown_url, output_path, quiet=False)

print(f"Archivo descargado en: {output_path}")
```

Ahora extraemos el contenido del archivo ZIP:

```python
# Definir la ruta del archivo ZIP descargado
output_path = '/content/dataset.zip'  # Donde se descargó el ZIP

# Crear directorio para el dataset
dataset_dir = '/content/dataset'
os.makedirs(dataset_dir, exist_ok=True)  # Crear el directorio si no existe

# Extraer el archivo ZIP
if os.path.exists(output_path):
    with zipfile.ZipFile(output_path, 'r') as zip_ref:
        print(f"Extrayendo archivos a {dataset_dir}...")
        zip_ref.extractall(dataset_dir)
    print(f"Extracción completada. Archivos disponibles en: {dataset_dir}")
else:
    print(f"Error: El archivo ZIP {output_path} no existe. Asegúrate de descargar primero el dataset.")
```

## 2. Explorar la estructura del dataset

Vamos a examinar la estructura de directorios del dataset:

```python
import glob

# Mostrar estructura de directorios
def show_directory_structure(directory, level=0):
    indent = "  " * level
    print(f"{indent}{os.path.basename(directory)}/")

    # Listar subdirectorios
    subdirs = [d for d in glob.glob(os.path.join(directory, "*")) if os.path.isdir(d)]
    for subdir in subdirs:
        show_directory_structure(subdir, level + 1)

    # Listar archivos (solo algunos si hay muchos)
    files = [f for f in glob.glob(os.path.join(directory, "*")) if os.path.isfile(f)]
    if len(files) > 5:
        for file in files[:3]:
            print(f"{indent}  {os.path.basename(file)}")
        print(f"{indent}  ... ({len(files) - 3} archivos más)")
    else:
        for file in files:
            print(f"{indent}  {os.path.basename(file)}")

show_directory_structure(dataset_dir)
```

## 3. Analizar el archivo de clases

Vamos a leer el archivo de clases para conocer las categorías del dataset:

```python
# Buscar archivo de clases (normalmente data.yaml)
yaml_files = glob.glob(os.path.join(dataset_dir, "**/*.yaml"), recursive=True)

import yaml

# Leer archivo YAML si existe
if yaml_files:
    with open(yaml_files[0], 'r') as f:
        yaml_data = yaml.safe_load(f)

    print("Información del dataset:")
    for key, value in yaml_data.items():
        print(f"{key}: {value}")
else:
    print("No se encontró archivo YAML de configuración")

    # Buscar si existe classes.txt como alternativa
    classes_files = glob.glob(os.path.join(dataset_dir, "**/classes.txt"), recursive=True)
    if classes_files:
        with open(classes_files[0], 'r') as f:
            classes = [line.strip() for line in f.readlines()]
        print("Clases encontradas:")
        for i, cls in enumerate(classes):
            print(f"{i}: {cls}")
```

## 4. Estadísticas del dataset
Vamos a analizar la distribución de las clases y otras estadísticas del dataset:

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Función para leer etiquetas en formato YOLO
def read_yolo_labels(label_file, img_width, img_height):
    boxes = []
    classes = []

    if not os.path.exists(label_file):
        return [], []

    with open(label_file, 'r') as f:
        for line in f.readlines():
            data = line.strip().split()
            if len(data) >= 5:
                class_id = int(data[0])
                # Formato YOLO: class_id, x_center, y_center, width, height (normalizados)
                x_center, y_center, width, height = map(float, data[1:5])

                # Convertir a coordenadas absolutas
                x1 = int((x_center - width/2) * img_width)
                y1 = int((y_center - height/2) * img_height)
                x2 = int((x_center + width/2) * img_width)
                y2 = int((y_center + height/2) * img_height)

                boxes.append([x1, y1, x2, y2])
                classes.append(class_id)

    return boxes, classes

# Encontrar carpetas de imágenes (train, val, test)
img_dirs = []
for subdir in ["train", "valid", "test"]:
    img_dir = os.path.join(dataset_dir, subdir, "images")
    if os.path.exists(img_dir):
        img_dirs.append((subdir, img_dir))

# Recopilar estadísticas
stats = {
    "split": [],
    "num_images": [],
    "class_counts": []
}

all_classes = []

for split_name, img_dir in img_dirs:
    images = glob.glob(os.path.join(img_dir, "*.jpg")) + glob.glob(os.path.join(img_dir, "*.png"))
    stats["split"].append(split_name)
    stats["num_images"].append(len(images))

    # Contar clases
    class_count = {}
    for img_path in images:
        # Construir ruta al archivo de etiquetas
        label_path = img_path.replace("images", "labels").rsplit(".", 1)[0] + ".txt"

        # Obtener dimensiones de la imagen
        from PIL import Image
        try:
            with Image.open(img_path) as img:
                img_width, img_height = img.size
        except Exception:
            continue

        # Leer etiquetas
        _, classes = read_yolo_labels(label_path, img_width, img_height)

        # Contar clases
        for cls in classes:
            if cls not in class_count:
                class_count[cls] = 0
            class_count[cls] += 1
            all_classes.append(cls)

    stats["class_counts"].append(class_count)

# Mostrar estadísticas básicas
for i in range(len(stats["split"])):
    print(f"\nConjunto '{stats['split'][i]}':")
    print(f"  - Número de imágenes: {stats['num_images'][i]}")
    print(f"  - Distribución de clases:")
    for cls, count in sorted(stats["class_counts"][i].items()):
        class_name = yaml_data['names'][cls] if 'names' in yaml_data else f"Clase {cls}"
        print(f"    - {class_name}: {count} instancias")

# Gráfico de distribución de clases
plt.figure(figsize=(12, 6))
all_classes_np = np.array(all_classes)
unique_classes = np.unique(all_classes_np)

class_names = []
for cls in unique_classes:
    if 'names' in yaml_data and cls < len(yaml_data['names']):
        class_names.append(yaml_data['names'][cls])
    else:
        class_names.append(f"Clase {cls}")

counts = [np.sum(all_classes_np == cls) for cls in unique_classes]

sns.barplot(x=class_names, y=counts)
plt.title('Distribución de clases en el dataset')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

## 5. Visualizar imágenes con sus anotaciones

Vamos a mostrar algunas imágenes de ejemplo con sus cajas delimitadoras:

```python
import cv2
import random
import matplotlib.pyplot as plt
from matplotlib import colors as mcolors  # Renombrado para evitar conflictos

# Generar colores aleatorios para las clases
def generate_colors(num_classes):
    # Usar colores distintivos para mejor visualización
    base_colors = list(mcolors.TABLEAU_COLORS.values())
    if num_classes <= len(base_colors):
        return base_colors[:num_classes]

    # Si necesitamos más colores, generar aleatoriamente
    np.random.seed(42)  # Para reproducibilidad
    return [tuple(np.random.random(3)) for _ in range(num_classes)]

# Función para dibujar cajas delimitadoras
def draw_boxes(image, boxes, classes, class_names, color_list):
    for box, cls_id in zip(boxes, classes):
        x1, y1, x2, y2 = box
        color = color_list[cls_id % len(color_list)]
        # Convertir color de matplotlib a formato BGR para OpenCV
        if isinstance(color, str):
            rgb = mcolors.to_rgb(color)  # Usando mcolors en lugar de colors
            color = (rgb[2]*255, rgb[1]*255, rgb[0]*255)
        elif len(color) == 3 and isinstance(color[0], float) and color[0] <= 1.0:
            color = (color[2]*255, color[1]*255, color[0]*255)

        # Dibujar rectángulo
        cv2.rectangle(image, (x1, y1), (x2, y2), color, 2)

        # Añadir etiqueta
        class_name = class_names[cls_id] if cls_id < len(class_names) else f"Clase {cls_id}"
        cv2.putText(image, class_name, (x1, y1-5), cv2.FONT_HERSHEY_SIMPLEX, 0.5, color, 2)

    return image

# Obtener nombres de clases
class_names = yaml_data['names'] if 'names' in yaml_data else [f"Clase {i}" for i in range(100)]

# Generar colores para las clases
class_colors = generate_colors(len(class_names))

# Mostrar algunas imágenes aleatorias
plt.figure(figsize=(15, 15))

# Buscar imágenes en los directorios
all_images = []
for _, img_dir in img_dirs:
    images = glob.glob(os.path.join(img_dir, "*.jpg")) + glob.glob(os.path.join(img_dir, "*.png"))
    all_images.extend(images)

# Seleccionar imágenes aleatorias
num_samples = min(9, len(all_images))
sample_images = random.sample(all_images, num_samples)

for i, img_path in enumerate(sample_images):
    # Leer imagen
    img = cv2.imread(img_path)
    img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

    # Obtener dimensiones
    img_height, img_width = img.shape[:2]

    # Construir ruta al archivo de etiquetas
    label_path = img_path.replace("images", "labels").rsplit(".", 1)[0] + ".txt"

    # Leer etiquetas
    boxes, classes = read_yolo_labels(label_path, img_width, img_height)

    # Dibujar cajas delimitadoras
    img_with_boxes = draw_boxes(img.copy(), boxes, classes, class_names, class_colors)

    # Mostrar imagen
    plt.subplot(3, 3, i + 1)
    plt.imshow(img_with_boxes)
    plt.title(f"Imagen {i+1}: {os.path.basename(img_path)}")
    plt.axis('off')

plt.tight_layout()
plt.show()
```

## 6. Análisis de Tamaños de Objetos

Vamos a analizar la distribución de tamaños de los objetos:

```python
# Recopilar información sobre tamaños de objetos
widths = []
heights = []
areas = []
aspect_ratios = []
class_ids = []

for _, img_dir in img_dirs:
    images = glob.glob(os.path.join(img_dir, "*.jpg")) + glob.glob(os.path.join(img_dir, "*.png"))

    for img_path in images:
        # Obtener dimensiones de la imagen
        try:
            with Image.open(img_path) as img:
                img_width, img_height = img.size
        except Exception:
            continue

        # Construir ruta al archivo de etiquetas
        label_path = img_path.replace("images", "labels").rsplit(".", 1)[0] + ".txt"

        # Leer etiquetas
        boxes, classes = read_yolo_labels(label_path, img_width, img_height)

        for box, cls in zip(boxes, classes):
            x1, y1, x2, y2 = box
            width = x2 - x1
            height = y2 - y1
            area = width * height
            aspect = width / height if height > 0 else 0

            widths.append(width)
            heights.append(height)
            areas.append(area)
            aspect_ratios.append(aspect)
            class_ids.append(cls)

# Crear un DataFrame para facilitar el análisis
import pandas as pd
df = pd.DataFrame({
    'width': widths,
    'height': heights,
    'area': areas,
    'aspect_ratio': aspect_ratios,
    'class': class_ids
})

# Visualizar distribución de tamaños
plt.figure(figsize=(15, 10))

plt.subplot(2, 2, 1)
sns.histplot(df['width'], bins=30)
plt.title('Distribución de Anchos de Objetos')
plt.xlabel('Ancho (píxeles)')

plt.subplot(2, 2, 2)
sns.histplot(df['height'], bins=30)
plt.title('Distribución de Altos de Objetos')
plt.xlabel('Alto (píxeles)')

plt.subplot(2, 2, 3)
sns.histplot(df['area'], bins=30)
plt.title('Distribución de Áreas de Objetos')
plt.xlabel('Área (píxeles²)')

plt.subplot(2, 2, 4)
sns.histplot(df['aspect_ratio'], bins=30)
plt.title('Distribución de Relaciones de Aspecto')
plt.xlabel('Relación de Aspecto (ancho/alto)')

plt.tight_layout()
plt.show()

# Visualizar tamaños por clase
plt.figure(figsize=(12, 8))
for cls in np.unique(df['class']):
    cls_data = df[df['class'] == cls]
    cls_name = class_names[cls] if cls < len(class_names) else f"Clase {cls}"
    plt.scatter(cls_data['width'], cls_data['height'], alpha=0.5, label=cls_name)

plt.title('Distribución de Tamaños por Clase')
plt.xlabel('Ancho (píxeles)')
plt.ylabel('Alto (píxeles)')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

## 7. Análisis de Ubicación de Objetos

Vamos a visualizar la ubicación de los objetos en las imágenes:

```python
# Recopilar información sobre ubicación de objetos
centers_x = []
centers_y = []
classes_for_centers = []

for _, img_dir in img_dirs:
    images = glob.glob(os.path.join(img_dir, "*.jpg")) + glob.glob(os.path.join(img_dir, "*.png"))

    for img_path in images:
        try:
            with Image.open(img_path) as img:
                img_width, img_height = img.size
        except Exception:
            continue

        # Construir ruta al archivo de etiquetas
        label_path = img_path.replace("images", "labels").rsplit(".", 1)[0] + ".txt"

        # Verificar si el archivo de etiquetas existe
        if not os.path.exists(label_path):
            # Intentar una alternativa: buscar el archivo de etiquetas con el mismo nombre base
            base_name = os.path.basename(img_path).split('.')[0]
            label_dir = os.path.dirname(img_path).replace("images", "labels")
            alternative_paths = glob.glob(os.path.join(label_dir, f"{base_name}*.txt"))

            if alternative_paths:
                label_path = alternative_paths[0]
            else:
                continue

        try:
            with open(label_path, 'r') as f:
                for line in f.readlines():
                    data = line.strip().split()
                    if len(data) >= 5:
                        class_id = int(data[0])
                        x_center, y_center = float(data[1]), float(data[2])

                        # Normalizar las coordenadas para todas las imágenes
                        centers_x.append(x_center)
                        centers_y.append(y_center)
                        classes_for_centers.append(class_id)
        except Exception as e:
            print(f"Error al leer {label_path}: {e}")
            continue

# Continuar con el resto del código solo si hay datos
if centers_x:
    # Crear un mapa de calor para visualizar la distribución de objetos
    plt.figure(figsize=(10, 8))

    # Crear una matriz para el mapa de calor
    heatmap_size = 100
    heatmap = np.zeros((heatmap_size, heatmap_size))

    # Llenar la matriz con las ubicaciones de los objetos
    for x, y in zip(centers_x, centers_y):
        # Convertir coordenadas normalizadas (0-1) a índices de matriz
        bin_x = min(int(x * heatmap_size), heatmap_size - 1)
        bin_y = min(int(y * heatmap_size), heatmap_size - 1)
        heatmap[bin_y, bin_x] += 1  # Incrementar el conteo en esa ubicación

    # Aplicar suavizado gaussiano para mejor visualización
    from scipy.ndimage import gaussian_filter
    heatmap_smooth = gaussian_filter(heatmap, sigma=1.5)

    # Visualizar el mapa de calor
    plt.imshow(heatmap_smooth, cmap='hot', origin='lower')
    plt.colorbar(label='Densidad de objetos')
    plt.title('Distribución Espacial de Objetos en el Dataset')
    plt.xlabel('Posición X normalizada')
    plt.ylabel('Posición Y normalizada')
    plt.tight_layout()
    plt.show()

    # Graficar la distribución por clase
    plt.figure(figsize=(12, 10))
    for cls in np.unique(classes_for_centers):
        # Obtener índices donde la clase es igual a cls
        indices = [i for i, c in enumerate(classes_for_centers) if c == cls]
        x_coords = [centers_x[i] for i in indices]
        y_coords = [centers_y[i] for i in indices]

        cls_name = class_names[cls] if cls < len(class_names) else f"Clase {cls}"
        plt.scatter(x_coords, y_coords, alpha=0.5, label=cls_name)

    plt.title('Ubicación de Objetos por Clase')
    plt.xlabel('Posición X normalizada')
    plt.ylabel('Posición Y normalizada')
    plt.legend()
    plt.grid(True, alpha=0.3)
    plt.xlim(0, 1)
    plt.ylim(0, 1)
    plt.gca().set_aspect('equal')
    plt.show()
else:
    print("No se encontraron datos suficientes para crear visualizaciones.")
```

## 8. Resumen del Dataset

Finalmente, vamos a generar un resumen del dataset:

```python
# Crear un resumen del dataset
print("=" * 50)
print("RESUMEN DEL DATASET DE PELEAS (YOLOv8)")
print("=" * 50)

# Información general
total_images = sum(stats["num_images"])
print(f"\nTotal de imágenes: {total_images}")

# Distribución por conjuntos
print("\nDistribución por conjuntos:")
for i, split in enumerate(stats["split"]):
    percentage = (stats["num_images"][i] / total_images) * 100
    print(f"  - {split}: {stats['num_images'][i]} imágenes ({percentage:.1f}%)")

# Clases y objetos
all_objects = len(widths)
print(f"\nTotal de objetos anotados: {all_objects}")

# Tabla de clases
print("\nDistribución de clases:")
class_counts = {}
for cls in class_ids:
    if cls not in class_counts:
        class_counts[cls] = 0
    class_counts[cls] += 1

for cls, count in sorted(class_counts.items()):
    cls_name = class_names[cls] if cls < len(class_names) else f"Clase {cls}"
    percentage = (count / all_objects) * 100
    print(f"  - {cls_name}: {count} objetos ({percentage:.1f}%)")

# Estadísticas de tamaño
print("\nEstadísticas de tamaño de objetos:")
print(f"  - Ancho promedio: {np.mean(widths):.1f} píxeles")
print(f"  - Alto promedio: {np.mean(heights):.1f} píxeles")
print(f"  - Área promedio: {np.mean(areas):.1f} píxeles²")
print(f"  - Relación de aspecto promedio: {np.mean(aspect_ratios):.2f}")
```
